# Creating a DataFrame 

```python
people = {
    "first": ["Shreyansh", "Jane", "John"],
    "last": ["Kumar", "Doe", "Doe"],
    "email": [
        "shreyan.kumar29@gmail.com",
        "Jane@gmail.com",
        "JohnDoe@gmail.com"
    ]
}
```
