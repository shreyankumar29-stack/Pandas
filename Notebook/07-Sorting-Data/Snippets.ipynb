```python
import pandas as pd
```

```python
people = {
    "first":["Shreyansh",'Jane','John', 'Adam'],
    "last":["Kumar",'Doe','Doe','Doe'],
    "email":["shre@gmail.com",'JaneDoe@gmail.com','JohnDoe@gmail.com','a@email.com']
}
```

```python
df = pd.DataFrame(people)
```

```python
df
```

# Sorting Single Columns

```python
# Example1: Sorting by last name
df.sort_values(by='last')
```

```python
# Example2: Sort these values in descendding order
df.sort_values(by='last', ascending=False)
```

# Sorting Multiple Columns

```python
# Example 1:
# 1. last name in descending order
# 2. if the last name contain duplicate then have to sort the first name
df.sort_values(by=['last','first'], ascending=False)
```

# Advanced Sorting Techniques

```python
# Ex1:
# 1. last name in descending order
# 2. first name in ascending order
df.sort_values(by=['last','first'], ascending=[False, True], inplace=True)
df
```

# Sorting index and series

```python
# Example: Sorting the index(sort index)
df.sort_index()
```

```python
# Example: Sorted Last Name
df['last'].sort_values()
```
