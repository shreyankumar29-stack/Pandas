```python
import pandas as pd
import numpy as np
```

```python
people = {
    'first': ['Corey', 'Jane', 'John', 'Chris', np.nan, None, 'NA'], 
    'last': ['Schafer', 'Doe', 'Doe', 'Schafer', np.nan, np.nan, 'Missing'], 
    'email': ['CoreyMSchafer@gmail.com', 'JaneDoe@email.com', 'JohnDoe@email.com', None, np.nan, 'Anonymous@email.com', 'NA'],
    'age': ['33', '55', '63', '36', None, None, 'Missing']
}
# nan stands for not a number value
```

```python
df = pd.DataFrame(people)

df.replace('NA', np.nan, inplace = True)
df.replace('Missing', np.nan, inplace = True)
```

```python
df
```

# Dropping Missing Values

```python
# Example: Removing the rows that have none values
# M1:
df.dropna()
```

```python
# M2:
df.dropna(axis='index', how='any')

# Explaination:
# 1. axis argument--->
#    a.)either set to index or set to column
#    b.)
#      I.)when set to index------>tell pandas that we want to drop na values when our rows are having missing values
#      II.)when set to columns--->if will drop the columns that have missing values
# 2. how argument---->
#    a.)how we want to drop these values
#    b.)uses for dropping a row or a column
#    c.)by default how='any'
#  3.df.dropna(axis='index', how='any')-----> 
#  here axis is set to index and how is set any it means that it will drop rows with any missing values
```

```python
# Example: Doesn't have an entire row that has missing values
df.dropna(axis='index', how='all')

# Explaination:
# df.dropna(axis='index', how='all')--> this will only drop rows when all of the values in that row are missing.
```

# Dropping subset of data

```python
# Example: Changing axis = 'index' to axis='columns'
df.dropna(axis='columns', how='all')
```

```python
df.dropna(axis='columns', how='any')
# an empty index
```

# Adavnced Dropping Techniques

```python
# Example: Suppose if thee dataframe doesn't contain first name and the last name and we need the email address and if they don't have an email address then we need to just drop those rows.
df.dropna(axis='index', how='any', subset=['email'])

# df.dropna(axis='index', how='all', subset=['email'])------------->gives the same result 
```

```python
# Example: in order for my data to be useful I need either their last name or their email address but I don't need both
df.dropna(axis='index', how='all', subset=['last', 'email'])
# Explaination:
# In this case the person is having the last name but not the email and vice versa because we have passed how='all' which means row to be dropped both the subset columns to be missing.
```
