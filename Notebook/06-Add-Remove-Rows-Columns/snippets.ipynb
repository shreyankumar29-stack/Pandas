````python
import pandas as pd
````

````python
people = {
    "first":["Shreyansh",'Jane','John'],
    "last":["Kumar",'Doe','Doe'],
    "email":["shre@gmail.com",'JaneDoe@gmail.com','JohnDoe@gmail.com']
}
````

````python
df = pd.DataFrame(people)
````

# Combining Columns

````python
# Example1: Combining first name and last name
df['full_name'] = df['first'] + ' ' + df['last']
df
````

# Removing Column

````python
# To delete multiple columns we need to pass list
df.drop(columns = ['first','last'], inplace =True)
df
````

# String Split 

````python
# Example2: If we want to split the full_name column
df['full_name'].str.split(' ')
````

# Expand

````python
# Example3: Expand this list in two different columns
df['full_name'].str.split(' ', expand=True)
````

# Multiple columns in DataFrames

````python
# Example 4:
df[['first', 'last']] = df['full_name'].str.split(' ', expand=True)
df
````

# Ignore Index

Error: `AttributeError: 'DataFrame' object has no attribute 'append'`
1. Reason for the Error

The error occurs because the `DataFrame.append()` method was removed in **Pandas 2.0**. Older Pandas tutorials use this method to add a single row or another DataFrame to an existing DataFrame. However, it is no longer available in modern Pandas versions.

For example, the following code produces an error:

```python
df.append({'first': 'Tony'})
```

The error occurs because Pandas cannot find the `append()` method in the DataFrame object.

2. Solution

The recommended alternative is the `pd.concat()` function, which combines two or more Pandas objects.

**Syntax:**

```python
pd.concat([df1, df2], ignore_index=True)
```

**Example:**

```python
new_row = pd.DataFrame([{'first': 'Tony'}])

df = pd.concat([df, new_row], ignore_index=True)
```

3. Explanation

* `pd.DataFrame([{'first': 'Tony'}])` creates a new DataFrame containing one row.
* `pd.concat([df, new_row])` combines the original DataFrame with the new row.
* `ignore_index=True` generates a new sequential index for the resulting DataFrame.
* `df =` assigns the combined DataFrame back to the variable `df`.

If the new row contains only the `first` column while the original DataFrame also contains `last` and `email`, Pandas fills the missing fields with `NaN`.

4. Important Note

The `append()` method was deprecated in Pandas 1.4 and removed in Pandas 2.0. Therefore, when following older Pandas tutorials, replace `df.append()` with `pd.concat()`.

**Remember:** `pd.concat()` returns a new DataFrame rather than modifying the original DataFrame directly. Assign the result back to a variable if you want to retain the combined data.

````python
new_row = pd.DataFrame([{'first': 'Tony'}])
df = pd.concat([df, new_row], ignore_index=True)
df
````
