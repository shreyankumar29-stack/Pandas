```python
# importing libraries
import pandas as pd
```

```python
people = {
    "first":["Shreyansh",'Jane','John'],
    "last":["Kumar",'Doe','Doe'],
    "email":["shre@gmail.com",'JaneDoe@gmail.com','JohnDoe@gmail.com']
}
```

```python
df = pd.DataFrame(people)
```

```python
df
```

```python
# Comparisons
df['last']=='Doe'
# We get a series with a bunch of boolean value
# These boolean value corresponds to our dataframes
# TRUE--->criteria met
# FALSE---> criteria not satisfied
```

```python
filt = (df['last']=='Doe')
```

```python
#Method1
df[filt]
```

```python
#method2
df[df['last']=='Doe']
```

```python
#method3
df.loc[filt] #we can still pass the specific columns that we want
```

```python
df.loc[filt, 'email']
```

```python
# &------>and
# |-------> OR
```

```python
# all of the rows that has the last name Doe and the first name is John
filt = (df['last']=='Doe') & (df['first']=='John')
```

```python
df.loc[filt, 'email']
```

```python
# the last name should be either ------- or -----------
filt = (df['last']=='Kumar') | (df['first']=='John')
df.loc[filt, 'email']
```

```python
# last name isn't Kumar or forst name isn't John
df.loc[~filt, 'email']
```
