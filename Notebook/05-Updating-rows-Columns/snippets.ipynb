Altering Data

```python
import pandas as pd
```

```python
people = {
    "first":["Shreyansh",'Jane','John'],
    "last":["Kumar",'Doe','Doe'],
    "email":["shre@gmail.com",'JaneDoe@gmail.com','JohnDoe@gmail.com']
}
```

```python
df = pd.DataFrame(people)
```

```python
df
```

```python
df.columns
```

```python
# changing column name(M1)
df.columns = ['first_name', 'last_name', 'email']
```

```python
df.columns
```

```python
df
```

```python
# uppercase to lowercase and vice versa
# colmnn name have spaces and want to replace them
# in the above two cases we use list comprehension
```

```python
# for example: I want to get all of the columns to be uppercase
df.columns = [x.lower() for x in df.columns]
df
```

```python
# For Example 2: Replace spaces with underscores
df.columns = df.columns.str.replace(' ', '_')
df
```

```python
# Example 3: We want to change some column (rename method and we will pass a dictionary)
df.rename(columns={
    'first_name':'first',
    'last_name': 'last'
},inplace=True)
df
```

```python
# Example 4: Updating data in rows
# Single value
df.loc[2] = ['John','Smith','JohnSmith@email.com']
df
```

```python
# Multiple value
df.loc[2, ['last', 'email']] = ['Doe', 'JohnDoe@email.com']
df
```

```python
# Example 5: To change only one value
df.loc[2, 'last'] = 'Smith'
df
```

```python
# another way 
df.at[2, 'last'] = 'Doe'
df
```

```python
# Common Mistake
filt=(df['email'] == 'JohnDoe@email.com')
df[filt]['last']='Smith'
df
```

```python
filt=(df['email'] == 'JohnDoe@email.com')
df.loc[filt, 'last']='Smith'
df
```

```python
# Example 6: Email Address to lowercase
df['email'] = df['email'].str.lower()
df
```

```python
# 4methods
#1. apply---->used for calling a function on our values(works on either dataframe or series object)
#2. map
#3. applymap
#4. replace
```

```python
df['email'].apply(len)
```

```python
def update_email(email):
    return email.upper()

df['email'].apply(update_email) # changes from lowercase to uppercase
#-----------------------------------------------------------------------
# Prints the table
df['email'] = df['email'].apply(update_email)
df
```
